Use the R kernel. The estimate mode fits missing model worlds and reuses completed compatible worlds. The summarize mode assembles saved model tables and shared-support summaries without estimating additional regressions. Both modes use the run selected by notebook 04.

```r
PROJECT <- normalizePath(getwd(), mustWork=TRUE)
MODE <- "estimate"  # estimate or summarize
stopifnot(MODE %in% c("estimate", "summarize"))
if (MODE == "summarize") {
  source(file.path(PROJECT, "benchmark_completed.R"))
  availability <- collect_completed_benchmarks(PROJECT)
} else {
  source(file.path(PROJECT, "benchmark_models.R"))
  run_all_benchmarks(PROJECT)
}
gc()
```
