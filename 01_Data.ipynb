Use the Python kernel and set data_root in config.json. Start Jupyter from the repository root. The workflow uses pandas with bounded Parquet I/O and publication years 1990-2024.

```python
from pathlib import Path
import sys
PROJECT=Path.cwd().resolve()
assert (PROJECT/'config.json').exists(),'Start Jupyter from the repository root.'
sys.path.insert(0,str(PROJECT))
import data as pipeline
from descriptive import describe
import pandas as pd
cfg=pipeline.load_config()
out=pipeline.output_root(cfg)
print('Input:',cfg['data_root'])
print('Output:',out)
```

```python
MODE='auto'  # auto or new
assert MODE in {'auto','new'}
ready=(out/'DATA_READY').exists() and any((out/'data').glob('*.parquet'))
if MODE=='new' or not ready:
    out=pipeline.run(cfg)
display(describe(cfg))
display(pd.read_csv(out/'tables/annual_sample.csv'))
```
