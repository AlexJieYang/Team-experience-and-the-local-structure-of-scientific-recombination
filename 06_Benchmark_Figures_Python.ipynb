```python
from pathlib import Path
import sys, importlib
import pandas as pd
from IPython.display import display,Image,FileLink
PROJECT=Path.cwd().resolve()
sys.path.insert(0,str(PROJECT))
import benchmark_profiles, benchmark_results, manuscript_export
for module in [benchmark_profiles,benchmark_results,manuscript_export]:importlib.reload(module)
c=benchmark_profiles.active_config(PROJECT)
print("Active run:",c["_out"])
```

```python
outputs=benchmark_results.run_all(c)
for out in outputs:
    print(out.name)
    display(pd.read_csv(out/"model_tables/null_global_comparisons.csv"))
    display(pd.read_csv(out/"model_tables/null_local_comparisons.csv"))
    display(pd.read_csv(out/"model_tables/null_profile_comparisons.csv"))
    for path in sorted((out/"figures/previews").glob("*.png")):
        display(Image(filename=str(path),width=1050))
for path in sorted((Path(c["_out"])/"comparison_figures/previews").glob("*.png")):
    display(Image(filename=str(path),width=1050))
```

```python
for out in outputs:
    insert=manuscript_export.export(c,out.name)
    print((insert/"benchmark_results.txt").read_text())
    display(FileLink(str(insert/"benchmark_results.tex")))
```
