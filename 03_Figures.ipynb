Use the Python kernel. This notebook reads exported model tables and plotting data. Set RESULTS_ROOT to the desired analysis directory. Figures 1-3 use two rows and three columns.

```python
from pathlib import Path
import json,sys
from IPython.display import display,Image,FileLink
PROJECT=Path.cwd().resolve()
sys.path.insert(0,str(PROJECT))
import importlib,figures_pnas
importlib.reload(figures_pnas)
from figures_pnas import render_all,load_settings
cfg=json.loads((PROJECT/'config.json').read_text(encoding='utf-8'))
RESULTS_ROOT=PROJECT/'outputs'/cfg['run_name']  # Set an absolute path to another completed analysis directory if needed.
assert (RESULTS_ROOT/'tables/profile_contrasts.csv').exists(),'RESULTS_ROOT must contain the exported tables directory.'
settings=load_settings(PROJECT/'figure_settings.json')
print('Results:',RESULTS_ROOT)
```

```python
report=render_all(RESULTS_ROOT,settings=settings)
print('Figures:',len(report['main']),'main,',len(report['supplement']),'SI')
display(FileLink(str(report['gallery'])))
display(FileLink(str(report['atlas'])))
```
